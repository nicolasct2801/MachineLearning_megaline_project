## comenzamos por importar las librerias que seran utilizadas

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import numpy as np
```

## leemos los datos despues de descargarlos

```python
data =  pd.read_csv('/datasets/users_behavior.csv')


print(data.head())
print(data.info())
print(data.describe())
```

aqui podemos observar que los datos ya han sido procesados y estan listos para utilizarse en un modelo

```python
#asignamos features y target
features = data.drop(['is_ultra'], axis=1)
target = data['is_ultra']
```

## ahora segmentamos los datos en target, train y test

<div class="alert alert-block alert-success"> 
<b>Comentario del Revisor #1</b> <a class="tocSkip"></a>

Excelente! Has dividido correctamente los datos en tres conjuntos: _train_, _validation_ y _test_. Has utilizado correctamente _train_test_split_ dos veces para dividir los datos en tres conjuntos (entrenamiento, validación y prueba). Este método es el más común para dividir los datos de manera aleatoria, asegurando que cada conjunto tenga una representación justa.

corregimos test_size en la siguiente celda

```python


features_train, features_valid, target_train, target_valid = train_test_split(features, target, test_size=0.4, random_state=54321)
features_valid, features_test, target_valid, target_test = train_test_split(features_valid, target_valid, test_size=0.5, random_state=54321)

print(f"Tamaño de features_train: {features_train.shape}")
print(f"Tamaño de target_train: {target_train.shape}")
print(f"Tamaño de features_valid: {features_valid.shape}")
print(f"Tamaño de target_valid: {target_valid.shape}")
print(f"Tamaño de features_test: {features_test.shape}")
print(f"Tamaño de target_test: {target_test.shape}")
```

## probamos regresion lineal

agregamos regresion lineal

```python
# Entrenamiento del modelo de Regresión Logística
logistic_model = LogisticRegression(random_state=54321, max_iter=1000)
logistic_model.fit(features_train, target_train)

# Evaluación en conjunto de validación
logistic_valid_predictions = logistic_model.predict(features_valid)
logistic_accuracy = accuracy_score(target_valid, logistic_valid_predictions)

print(f'Precisión de Regresión Logística (Validación): {logistic_accuracy:.4f}')
```

## probamos con arbol de decision

agregamos arbol de desicion

```python
tree_model = DecisionTreeClassifier(random_state=54321, max_depth=10)
tree_model.fit(features_train, target_train)

# Evaluación en conjunto de validación
tree_valid_predictions = tree_model.predict(features_valid)
tree_accuracy = accuracy_score(target_valid, tree_valid_predictions)

print(f'Precisión del Árbol de Decisión (Validación): {tree_accuracy:.4f}')
```

## probamos RandomForestClassifier 
porque es un modelo muy sólido que ofrece buenos resultados para clasificación sin necesidad de un ajuste complejo, y es menos propenso a sobreajustarse y a ser afectado por datos ruidosos o correlacionados.

<div class="alert alert-block alert-success"> 
<b>Comentario del Revisor #1</b> <a class="tocSkip"></a>

Has entrenado el modelo únicamente con el conjunto de entrenamiento, lo que es exactamente lo que se debe hacer. Este conjunto contiene los datos con los que el modelo aprende los patrones, y no debe utilizar otros conjuntos durante el entrenamiento.

Luego, has validado correctamente el modelo utilizando solo el conjunto de validación. Esta es una práctica esencial para ajustar los hiperparámetros y comparar diferentes modelos sin introducir sesgos.


</div>

```python

model = RandomForestClassifier(random_state=54321)

# Entrenar el modelo
model.fit(features_train, target_train)

# Predecir en el conjunto de validación
target_valid_pred = model.predict(features_valid)

# Evaluar la precisión en el conjunto de validación
accuracy_valid = accuracy_score(target_valid, target_valid_pred)
print(f'Precisión en la validación: {accuracy_valid:.4f}')
```

la precisión en la validación es 0.7823, lo que significa que el 78.23% de las predicciones realizadas por el modelo en el conjunto de validación fueron correctas. Nos dio lo mismo que DecisionTreeClassifier y decidimos quedarnos con RandomForestClassifier

<div class="alert alert-block alert-info"> <b>Utilizamos un bucle for para iterar e posibles valores de hiperparametros y escoger el mejor modelo.</b> <a class="tocSkip"></a> </div>

```python
best_score_tree = 0
best_params_tree = {}

for depth in range(1, 21):
    for criterion in ['gini', 'entropy']:
        tree_model = DecisionTreeClassifier(max_depth=depth, criterion=criterion, random_state=54321)
        tree_model.fit(features_train, target_train)
        predictions_valid = tree_model.predict(features_valid)
        score = accuracy_score(target_valid, predictions_valid)
        
        if score > best_score_tree:
            best_score_tree = score
            best_params_tree = {'max_depth': depth, 'criterion': criterion}

print(f"Mejores parámetros para DecisionTreeClassifier: {best_params_tree}")
print(f"Mejor puntuación en validación: {best_score_tree:.4f}")

# Ajuste de hiperparámetros para RandomForestClassifier
best_score_forest = 0
best_params_forest = {}

for n_estimators in range(10, 201, 10):  # Probar de 10 a 100 árboles
    for depth in range(5, 21, 5):        # Probar profundidades de 5 a 20
        model = RandomForestClassifier(n_estimators=n_estimators, max_depth=depth, random_state=54321)
        model.fit(features_train, target_train)
        predictions_valid = model.predict(features_valid)
        score = accuracy_score(target_valid, predictions_valid)
        
        if score > best_score_forest:
            best_score_forest = score
            best_params_forest = {'n_estimators': n_estimators, 'max_depth': depth}

print(f"Mejores parámetros para RandomForestClassifier: {best_params_forest}")
print(f"Mejor puntuación en validación: {best_score_forest:.4f}")
```

## nos quedamos con RandomForestClassifier

```python
model = RandomForestClassifier(n_estimators=10, max_depth=10, random_state=54321)

# Reentrenar el modelo
model.fit(features_train, target_train)

# Predecir en el conjunto de validación
target_valid_pred = model.predict(features_valid)

# Evaluar la precisión en el conjunto de validación
accuracy_valid = accuracy_score(target_valid, target_valid_pred)
print(f'Precisión ajustada en la validación: {accuracy_valid:.4f}')
```

El modelo ha sido ajustado y ha mejorado ligeramente su precisión en la validación, alcanzando un 80.25% de aciertos
